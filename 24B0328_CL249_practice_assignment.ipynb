# Question 1: List Filtering & Data Processing

```python
titles = ["Dune", "The Hobbit", "1984", "Foundation", "Hamlet"]

filtered = []
for t in titles:
    if len(t) > 5:
        filtered.append(t)

print("Original list:", titles)
print("Filtered list:", filtered)
print("Total filtered titles:", len(filtered))
```

# Question 2: Array Operations & Vectorization with NumPy

```python
import numpy as np

celsius_readings = [18.5, 19.0, 21.2, 24.5, 26.8, 27.1, 25.4, 22.0]
celsius_arr = np.array(celsius_readings)

fahrenheit_arr = (celsius_arr * 1.8) + 32

avg_f = np.mean(fahrenheit_arr)
max_f = np.max(fahrenheit_arr)
above_75 = fahrenheit_arr[fahrenheit_arr > 75]

print("Celsius readings:", celsius_arr)
print("Fahrenheit readings:", fahrenheit_arr)
print("Average temperature (F):", avg_f)
print("Maximum temperature (F):", max_f)
print("Readings above 75F:", above_75)
```

# Question 3: Pseudocode Conversion & Function Logic

```python
def calculate_checkout(customer_tier, total_amount):
    if customer_tier == "Gold":
        discount_rate = 0.20
    elif customer_tier == "Silver":
        discount_rate = 0.10
    else:
        discount_rate = 0.05

    if total_amount > 100.0:
        discount_rate = discount_rate + 0.05

    final_price = total_amount * (1.0 - discount_rate)
    return final_price

result1 = calculate_checkout("Gold", 120.0)
result2 = calculate_checkout("Standard", 50.0)

print("Final total: ${:.2f}".format(result1))
print("Final total: ${:.2f}".format(result2))
```

# Question 4: Random Growth Sequence Simulation

```python
import numpy as np
import matplotlib.pyplot as plt
import random

initial_value = 100
steps = 30
sequence = [initial_value]

current_value = initial_value
for i in range(steps):
    random_factor = random.uniform(0.5, 1.5)
    current_value = current_value * random_factor
    sequence.append(current_value)

sequence = np.array(sequence)

plt.plot(sequence)
plt.xlabel("Iteration")
plt.ylabel("Value")
plt.title("Random Growth Sequence")
plt.show()
```

# Question 5: Gauss-Seidel Method for Linear Equations

```python
a11, a12, a13, b1 = 10, 1, -1, 11
a21, a22, a23, b2 = 1, 10, 1, 12
a31, a32, a33, b3 = 2, -1, 10, 21

x, y, z = 0.0, 0.0, 0.0
tolerance = 0.0001
max_iterations = 50

for iteration in range(1, max_iterations + 1):
    x_old, y_old, z_old = x, y, z

    x = (b1 - a12 * y - a13 * z) / a11
    y = (b2 - a21 * x - a23 * z) / a22
    z = (b3 - a31 * x - a32 * y) / a33

    if abs(x - x_old) < tolerance and abs(y - y_old) < tolerance and abs(z - z_old) < tolerance:
        print("Converged in iteration:", iteration)
        break

print("x =", round(x, 4))
print("y =", round(y, 4))
print("z =", round(z, 4))
print("Total iterations:", iteration)
```
